Sets
- Contains unordered collections of unique items. They are defined much like lists and tuples, except they use the curly brackets of dictionaries:

```python
primes = {2, 3, 5, 7}
odds = {1, 3, 5, 7, 9}
```

- Union, intersection, difference, symmetric difference, and others. Python's sets have all of these operations built-in, via methods or operators.

```python
# union: items appearing in either
primes | odds      # with an operator
primes.union(odds) # equivalently with a method
```

```python
# intersection: items appearing in both
primes & odds             # with an operator
primes.intersection(odds) # equivalently with a method
```

```python
# difference: items in primes but not in odds
primes - odds           # with an operator
primes.difference(odds) # equivalently with a method
```

```python
# symmetric difference: items appearing in only one set
primes ^ odds                     # with an operator
primes.symmetric_difference(odds) # equivalently with a method
```

Useful Iterators

- Enumerate
  - Often you need to iterate not only the values in an array, but also keep track of the index.
  - It essentially wraps an iterable (like a list or string) and returns an iterator that yields pairs of data: (index, item).

```python
L = [2, 4, 6, 8, 10]
for i in range(len(L)):
    print(i, L[i])
```

- Cleaner Syntax using the enumerate iterator:

```python
for i, val in enumerate(L): #i, val--> (index, value), #enumerate(L)-->function takes the iterable sequence L (which could be a list, tuple, string, etc.) and turns it into a stream of (index, value) pairs
    print(i, val)
```

- Zip
  - Other times, you may have multiple lists that you want to iterate over simultaneously.
  - Zip iterator zips together iterables
  - Any number of iterables can be zipped together, and if they are different lengths, the shortest will determine the length of the zip

```python
#for loop iterates through the pairs generated by zip(). In each iteration, the tuple is automatically unpacked:
#element from list L is assigned to the variable lval
#element from list R is assigned to the variable rval
# since both lists have a length of 5, the loop will run 5 times

L = [2, 4, 6, 8, 10]
R = [3, 6, 9, 12, 15]
for lval, rval in zip(L, R):
    print(lval, rval)
```

- Map and Filter
  - The map iterator takes a function and applies it to the values in an iterator:

```python
# find the first 10 square numbers
# A lambda function in Python creates a small, anonymous function—meaning it doesn't have a formal name defined with def. Think of it as a quick, inline function shortcut.
# The map() function takes the square function and applies it to every item in the range(10) iterable. It returns a map object (an iterator) that yields the results one by one:
# syntax: map(function, iterable)
square = lambda x: x ** 2 #Create a function that takes a number, x, and gives back its square
for val in map(square, range(10)):
    print(val, end=' ')
```

- The filter iterator looks similar, except it only passes-through values for which the filter function evaluates to True:

```python
# find values up to 10 for which x % 2 is zero
is_even = lambda x: x % 2 == 0
for val in filter(is_even, range(10)):
    print(val, end=' ')
```

- The map and filter functions, along with the reduce function (which lives in Python's functools module) are fundamental components of the functional programming style, which, while not a dominant programming style in the Python world, has its outspoken proponents

Iterators as Function Arguments
-  *args and **kwargs can be used to pass sequences and dictionaries to functions. It turns out that the *args syntax works not just with sequences, but with any iterator:
  - *args and **kwargs are special syntax in Python used within function definitions to allow a function to accept an arbitrary number of arguments. They are fundamentally about packing and unpacking arguments.
  - *args (Arbitrary Positional Arguments): The single asterisk (*) tells Python to gather any extra positional arguments passed to the function into a tuple.
  - **kwargs (Arbitrary Keyword Arguments): The double asterisk (**) tells Python to gather any extra keyword arguments (arguments passed with a name, like name="Alice") into a dictionary.
  - The point of using *args and **kwargs is to make your functions flexible by allowing them to accept an arbitrary (unspecified) number of arguments.

```python
print(*range(10)) #The individual elements of the range, separated by spaces will be printed out.
```

```python
print(*map(lambda x: x ** 2, range(10))) #map function must have at least two arguments
```

- The key is that zip() can zip-together any number of iterators or sequences.

```python
L1 = (1, 2, 3, 4)
L2 = ('a', 'b', 'c', 'd')
```

```python
z = zip(L1, L2) #Creates an iterator of pairs.
print(*z) #Unpacks the iterator elements as separate arguments to print.
```

```python
z = zip(L1, L2)
new_L1, new_L2 = zip(*z)
print(new_L1, new_L2)
```

Specialized Iterators: itertools
- A powerful module in Python's standard library that provides a collection of fast, memory-efficient tools for working with iterators.
- It allows you to build complex iteration patterns using small, efficient, composable functions.

```python
from itertools import permutations #intertools.permutations function iterates over all permutations of a sequence
p = permutations(range(3)) #length argument (r) was ommitted from the function call so defaults to r=3 (length of 3)
print(*p)
```

- Similarly, the itertools.combinations function iterates over all unique combinations of N values within a list:

```python
from itertools import combinations
c = combinations(range(4), 2) #range(4)- input elements {0, 1, 2, 3} #2- combinations of the length 2 (pairs)
print(*c)
```

- Product iterator, which iterates over all sets of pairs between two or more iterables:

```python
from itertools import product
p = product('ab', range(3)) #first iterable is the string 'ab' which acts as the sequence of characters #second iterable is range(3)
print(*p)
```

- Anyway, once you have a sorted list, you can use a technique binary search to find an item in the list pretty quickly.
- The basic idea is that you start looking at the middle of the list, and compare the middle entry with what you are looking for.
- Bisect module in Python provides functions for maintaining a list in sorted order without having to sort the entire list repeatedly. It uses an efficient search algorithm called binary search.

```python
import bisect
```

```python
bisect.bisect_left(my_ints, 5)
```

- Why left?
- Because the bisect functions are named assuming you will use the results to insert a new item, and inserting to the left of existing items means taking the existing item's position.

```python
bisect.bisect_left(my_ints, 4)
```

- The key difference between Python's built-in sort() method and the sorted() function lies in where and how they modify the data.
  - list.sort()(Method)
    - in-place: modifies the original list directly
    - returns None
    - list-only: can only be called on a list object
  - sorted(iterable)(Function)
    - non- destructive: creates and returns a new sorted list
    - returns the newly sorted list
    - universal: works on any iterable (lists, tuples, strings, dictionaries, generators, etc.)

- The bisect_left function is telling you where the item was found, or where it would be inserted.
- You need to check if it it was actually found.
- Moving on, binary search lets you find an item in the list quickly if the list is sorted, but you'll need to keep sorting the lists between adding new entries and searching.
- Python's sort is optimized for these repeated sort cases, but the total work to sort the list can still be quadratic if you sort after every item is added.
- Dictionaries, which we will cover shortly, will be even faster if we know exactly what we are looking for, and they don't need repeated sorting.
- So, lists are convenient for holding and looping over data, but not great for searching if it keeps changing.

Python Dictionaries
- Lists can hold, filter, and transform data, but they are not great for searching
- Python dictionaries are great for finding items if you know exactly what you are looking for
- The main idea for dictionaries is that you need to choose an immutable key.
- The key is how we specify exactly what we are looking for.
- And it needs to be immutable, meaning that you can not change it, because the data structures rely on that immutability to be able to find the key later.
- If you change the key, the data structures might look in the wrong place and have trouble finding the data you added before.
- Technically, what Python is checking is whether it can "hash" the key.
  - Hashing is the fundamental mechanism that allows Python dictionaries to work so incredibly fast

```python
hash(3)
```

```python
hash("3")
```

```python
hash([3])
```

- Lists do not work because you could change the list.
- You could add more data, or change the data already in it.

```python
hash((3,))
```

- But tuples do work as long as everything in the tuple can be hashed.
- The dictionary data structure uses those hash values to decide where to put data, and to find it later.
- So they need to be consistent or the dictionary will look in the wrong place and won't find your data.
- That's why the hash keys must be immutable.
