Dictionaries
- Extremely flexible mappings of keys to values, and form the basis of much of Python's internal implementation. They can be created via a comma-separated list of key:value pairs within curly braces:

```python
numbers = {'one':1, 'two':2, 'three':3}
```

- Items are accessed and set via the indexing syntax used for lists and tuples, except here the index is not a zero-based order but valid key in the dictionary:

```python
# Access a value via the key
numbers['two']
```

- New items can be added to the dictionary using indexing as well:

```python
# Set a new key:value pair
numbers['ninety'] = 90
print(numbers)
```

- Keep in mind that dictionaries do not maintain any sense of order for the input parameters; this is by design. This lack of ordering allows dictionaries to be implemented very efficiently, so that random element access is very fast, regardless of the size of the dictionary

Sets
- Contains unordered collections of unique items. They are defined much like lists and tuples, except they use the curly brackets of dictionaries:

```python
primes = {2, 3, 5, 7}
odds = {1, 3, 5, 7, 9}
```

- Union, intersection, difference, symmetric difference, and others. Python's sets have all of these operations built-in, via methods or operators.

```python
# union: items appearing in either
primes | odds      # with an operator
primes.union(odds) # equivalently with a method
```

```python
# intersection: items appearing in both
primes & odds             # with an operator
primes.intersection(odds) # equivalently with a method
```

```python
# difference: items in primes but not in odds
primes - odds           # with an operator
primes.difference(odds) # equivalently with a method
```

```python
# symmetric difference: items appearing in only one set
primes ^ odds                     # with an operator
primes.symmetric_difference(odds) # equivalently with a method
```

Identity and Membership Operators
- Like and, or, and not, Python also contains prose-like operators to check for identity and membership. They are the following:

Operator	Description
- a is b-->	True if a and b are identical objects
- a is not b-->	True if a and b are not identical objects
- a in b-->	True if a is a member of b
- a not in b-->	True if a is not a member of b


Identity Operators "is" and "is not"
- The identity operators, "is" and "is not" check for object identity. Object identity is different than equality, as we can see here:

```python
a = [1, 2, 3]
b = [1, 2, 3]
```

```python
a == b
```

```python
a is b
```

```python
a is not b
```

```python
#Example of identical objects
a = [1, 2, 3]
b = a
a is b
```

- The difference between the two cases here is that in the first, a and b point to different objects, while in the second they point to the same object.

Membership Operators
- Check for membership within compound objects

```python
1 in [1, 2, 3]
```

```python
2 not in [1, 2, 3]
```

Iterators
- An object that allows you to traverse, or go through, a sequence of elements one at a time.
- Think of it like moving through a list or array one step at a time using a pointer, rather than loading the whole thing into memory at once

```python
for i in range(10): #for i: defines the loop variable #i in range(10): iterable
    print(i, end=' ')
```

- Range is not a list, but is something called an iterator

Iterating over Lists


```python
for value in [2, 4, 6, 8, 10]: #"for x in y" syntax allows us to repeat some operation for each value in the list
    # do some operation
    print(value + 1, end=' ')
```

- When you write something like "for val in L", the Python interpreter checks whether it has an iterator interface, which you can check yourself with the built-in iter function:

```python
iter([2, 4, 6, 8, 10])
```

- It is this iterator object that provides the functionality required by the for loop. The iter object is a container that gives you access to the next object for as long as it's valid, which can be seen with the built-in function next:

```python
I = iter([2, 4, 6, 8, 10])
```

```python
print(next(I))
```

```python
print(next(I))
```

```python
print(next(I))
```

- Allows Python to treat things as lists that are not actually lists.

range(): A List is Not Always a List
- Most common example of this indirect iteration is the range() function, which returns not a list, but a special range() object:



```python
range(10)
```

- Range, like a list, exposes an iterator:

```python
iter(range(10))
```

- So Python knows to treat it as if it's a list:

```python
for i in range(10):
    print(i, end=' ')
```

- The benefit of the iterator indirection is that the full list is never explicitly created!

```python
N = 10 ** 12
for i in range(N):
    if i >= 10: break
    print(i, end=', ')
```
