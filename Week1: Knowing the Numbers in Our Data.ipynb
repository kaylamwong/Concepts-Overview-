- Numbers, strings, and other simple types are immutable: you can't change their value – you can only change what values the variables point to.

```python
x = 10 
y = x 
x +=5 # add 5 to x's value, and assign it to x 
print("x =", x)
print("y=", y)
```

- When we call x += 5, we are not modifying the value of the 10 object pointed to by x; we are rather changing the variable x so that it points to a new integer object with value 15. For this reason, the value of y is not affected by the operation.

```python
# Defining Functions 

def fibonacci(N): 
    L = [] 
    a, b = 0, 1 
    while len (L) < N: 
        a, b = b, a + b 
        L.append(a)
    return L 
```

```python
fibonacci(10)
```

```python
def real_imag_conj(val): 
    return val.real, val.imag, val.conjugate() #multiple return values are put in a tuple, indicated by commas

r, i, c= real_imag_conj(3 + 4j)
print(r, i, c)
```

- Default values for arguments allow you to specify a default value for a parameter in a function definition. If the caller does not provide a value for that parameter, the default value is used instead. This makes functions more flexible and convenient to call.

```python
def fibonacci(N, a=0, b=1):
    L = []
    while len(L) < N:
        a, b = b, a + b
        L.append(a)
    return L
```

```python
fibonacci(10)
```

```python
fibonacci(10, 0, 2)
```

```python
fibonacci(10, b=3, a=1)
```

- *args and **kwargs (aka arguments and keyword arguments): Flexible Arguments 
    - Are special syntax used to allow your function to accept an arbitrary (variable) number of arguments
    - Imagine you're defining a function, but you don't know exactly how many items the user will give it. These two tools are Python's way of saying, "Just give me whatever you have."
    - *args syntax lets the function accept any number of extra items that aren't specifically named 
        - Python takes al those extra items and packs them into a single tuple (an organized container) called args 
    - **kwargs syntax lets the function accept any number of extra pieces of information that have names attached to them 
        - Python takes those extra name- value pairs and packs them into a single dictionary (a collection of keys and values) called kwargs 
    - The operative difference is the asterisk characters: a single * before a variable means "expand this as a sequence", while a double ** before a variable means "expand this as a dictionary".

```python
# Simple *args example 
#calculate(10, 20, 30)--> (10, 20, 30) (a tuple)

#Simple **kwargs example 
#profile(age=30, city="NY")--> ('age': 30, 'city': 'NY') (a dictionary)


```

```python
def catch_all(*args, **kwargs): 
    print("args =", args)
    print("kwargs=", kwargs)
```

```python
catch_all(1, 2, 3, a=4, b=5)
```

```python
catch_all('a', keyword=2)
```

```python
inputs = (1, 2, 3)
keywords= {'pi': 3.14}

catch_all(*inputs, **keywords)
```

- Dictionaries are not orderable: we need a way to tell the function how to sort our data. We can do this by specifying the key function, a function which given an item returns the sorting key for that item:

```python
# sort alphabetically by first name
sorted(data, key=lambda item: item['first'])
```

```python
# sort by year of birth
sorted(data, key=lambda item: item['YOB'])
```
