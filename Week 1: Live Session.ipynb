- Arithmetic Operations

```python
x = 3 
y = 14 
mult = x * y #multiplication 
div = x / y #division (floating point division, real numbers)
div2 = x // y #integer division 
mod = x % y #modulo: remainder from integer division 

print("Results: ")
print("multiplication:", mult)
print("floating point division:", div)
print("integer division:", div2)
print("remainder from integer division:", mod)
```

- Functions 
    - package up some statements of code to reuse them later 
    - use "parameters" 
    - return a result 

```python
def mult_three(x): 
    '''multiply the parameter
    x by 3 and return the result'''
    result = x * 3 
    return result #copy the value out of the function 
# When Python encounters a return statement, the function immediately stops executing, even if there is more code after it.
```

- The point of the return statement is crucial: it's how a function sends a result back to the part of the code that called it. Without return, a function might perform calculations, but it effectively keeps the result hidden inside itself.
