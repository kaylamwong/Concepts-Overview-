- You should never use the direct equality operator (==) to compare floating-point numbers. Instead, you should check if the two numbers are within a small, acceptable tolerance, known as epsilon.

```python
import math

# Checks if the values are close enough to be considered equal
result = math.isclose(1.1 + 2.2, 3.3) 

print(result) 
```

- Floating Point Number: scientific notation in base 2 (binary)
    - Computers use floating point to efficiently store a wide range of numbers, from the size of the universe to the size of an atom
    - Structure: They store the significant digits (the numbers that matter) and the position of the decimal (or binary) point separately.
    - The core reason for errors lies in the difference between the human Base 10 (decimal) system and the computer's Base 2 (binary) system.

- DON'T: Use if a + b == c: for floating-point numbers.

- DO: Use a tolerance check like math.isclose(a + b, c) to safely determine if the numbers are close enough for all practical purposes.

- Integers: most basic numerical type is the integer. Any number without a decimal point is an integer:

```python
x = 1 
type(x)
```

- Floating- Point Numbers can store fractional numbers. They can be defined either in standard decimal notation, or in exponential notation:
    - Precision is limited, which can cause equality tests to be unstable
    - Floating-point arithmetic is approximate, and never rely on exact equality tests with floating-point values.

```python
x = 0.000005
y = 5e-6
print(x == y)
```

```python
float(1)
```

```python
0.1 + 0.2 == 0.3
```

```python
print("0.1 = {0:.17f}".format(0.1))
print("0.2 = {0:.17f}".format(0.2))
print("0.3 = {0:.17f}".format(0.3))
```
