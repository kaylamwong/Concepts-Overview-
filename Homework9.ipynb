### Problem 15

Write a function `p15` that takes in three filenames as input; a file name of device orders, a file name for device components, and a file name of component costs; and returns the total cost.

* The orders file contains two columns -- `device` and `device_quantity`.
* The components file contains three columns -- `device`, `component`, and `component_quantity`.
* The costs file contains two columns -- `component` and `component_cost`.

```python
from google.colab import files
```

```python
 uploaded = files.upload()
```

```python
# YOUR CHANGES HERE

import pandas as pd

def p15(orders_filename, components_filename, costs_filename):
    # --- 1. Load the three dataframes with explicit column names ---
    orders = pd.read_csv(orders_filename, sep='\t', names=['device', 'device_quantity'], header=None)
    components = pd.read_csv(components_filename, sep='\t', names=['device', 'component', 'component_quantity'], header=None)
    costs = pd.read_csv(costs_filename, sep='\t', names=['component', 'component_cost'], header=None)

    # **CRITICAL FIXES: 1. Data Types and 2. Cleaning Merge Keys**

    # Convert numeric columns to float
    orders['device_quantity'] = orders['device_quantity'].astype(float)
    components['component_quantity'] = components['component_quantity'].astype(float)
    costs['component_cost'] = costs['component_cost'].astype(float)

    # Clean merge keys: strip whitespace and convert to lower case for consistent matching
    orders['device'] = orders['device'].astype(str).str.strip().str.lower()
    components['device'] = components['device'].astype(str).str.strip().str.lower()
    components['component'] = components['component'].astype(str).str.strip().str.lower()
    costs['component'] = costs['component'].astype(str).str.strip().str.lower()

    # --- STEP 1: Calculate Unit Cost per Device ---

    # 1a. Merge components and costs (on 'component')
    comp_details = pd.merge(components, costs, on='component', how='inner')

    # 1b. Calculate the total cost for that specific component in ONE device
    comp_details['total_comp_cost'] = comp_details['component_cost'] * comp_details['component_quantity']

    # 1c. Group by device and sum to get the final DEVICE_UNIT_COST.
    device_unit_cost = comp_details.groupby('device')['total_comp_cost'].sum().reset_index()
    device_unit_cost.rename(columns={'total_comp_cost': 'device_unit_cost'}, inplace=True)

    # --- STEP 2: Calculate Total Order Cost ---

    # 2a. Merge orders with the calculated device unit costs (on 'device').
    # This merge now works correctly because the keys have been cleaned.
    final_data = pd.merge(orders, device_unit_cost, on='device', how='inner')

    # 2b. Calculate the total cost for each line item order (Multiplication works now due to type conversion)
    final_data['total_order_cost'] = (
        final_data['device_quantity'] * final_data['device_unit_cost']
    )

    # 2c. Return the grand total sum across all orders
    total_cost = final_data['total_order_cost'].sum()
    return total_cost
```

### Problem 16

Write a function `p16` that takes in a list of filenames, reads all of them, and returns all of their rows in a single data frame in the same order as the input files and the order they were specified.
You may assume that all of the files have TSV format and the same columns.

Hint: you've seen a simpler version of this behavior loading penguin data.

```python
from google.colab import files
```

```python
 uploaded = files.upload()
```

```python
# YOUR CHANGES HERE

def p16(filenames):
    """
    Reads all rows from a list of TSV files and returns them in a single DataFrame,
    preserving the order of the input files.
    """

    # 1. Create an empty list to hold the DataFrames
    df_list = []

    # 2. Loop through each filename in the input list
    for file in filenames:
        # Read the TSV file, specifying the tab separator
        # Assume files have a header row, so no 'header=None' is needed
        df = pd.read_csv(file, sep='\t')

        # Add the read DataFrame to our list
        df_list.append(df)

    # 3. Concatenate all DataFrames in the list
    # pd.concat joins them one after another (vertically)
    final_df = pd.concat(df_list, ignore_index=True)

    # 4. Return the combined DataFrame
    return final_df
```

### Problem 20

Write a function `p20` that takes in a filename for components of a single device, a list of filenames where each file has a different set of device costs, and returns a data frame of costs by filename.
The output should have two columns, "filename" and "cost".

Hint: you have already done similar analysis for multiple devices and designs. The main difference here is that the cost data is spread out across multiple files, and you need to report analysis results per file. How should you load the cost data while keeping it associated to the right files?

```python
from google.colab import files
```

```python
# YOUR CHANGES HERE

def p20(components_filename, cost_filenames):
    # 1. Load the single components file
    # This file links 'device' and 'component' to 'component_quantity'
    # It is read as a tab-separated file.
    components = pd.read_csv(components_filename, sep='\t')

    # 2. Initialize a list to hold the results (filename and cost)
    results_list = []

    # 3. Loop through each cost file provided in the list
    for filename in cost_filenames:

        # a. Read the current cost file
        # The cost file contains two columns: 'component' and 'component_cost'.
        # We need to handle header issues as previous problems for robustness.
        try:
            # Assumes standard header is present
            costs = pd.read_csv(filename, sep='\t')
        except:
            # If standard read with assumed header fails, try again without headers
            # Note: This might not be strictly necessary if the files are consistent,
            # but it demonstrates robustness. The previous problem hint mentioned
            # handling this. Given the file names, the standard read should work.
            costs = pd.read_csv(filename, sep='\t', names=['component', 'component_cost'], header=None)

        # b. Merge components with the current costs to find the cost per component
        comp_details = pd.merge(components, costs, on='component', how='inner')

        # c. FIX! Convert quantity and cost columns to numeric types.
        # This is essential to prevent 'total_cost' calculation from failing due to 'object' type.
        comp_details['component_quantity'] = pd.to_numeric(comp_details['component_quantity'], errors='coerce')
        comp_details['component_cost'] = pd.to_numeric(comp_details['component_cost'], errors='coerce')

        # d. Calculate the total cost for that specific component in ONE device
        # Formula: total_cost = quantity * unit_cost
        comp_details['total_cost_comp'] = comp_details['component_quantity'] * comp_details['component_cost']

        # e. Calculate the total unit cost for the final device (sum of all component costs)
        # Note: Since the problem implies a single device's components are listed,
        # we can just sum the 'total_cost_comp' column across the entire DataFrame.
        total_cost = comp_details['total_cost_comp'].sum()

        # f. Store the Result as a dictionary
        # CRITICAL CHANGE: The key is changed from "total_cost" to **"cost"** # to match the expected output column name for the test.
        results_list.append({
            "filename": filename,
            "cost": total_cost  # <-- **CORRECTED COLUMN NAME**
        })

    # 4. Convert the list of results into the final DataFrame
    final_df = pd.DataFrame(results_list)

    return final_df
```
